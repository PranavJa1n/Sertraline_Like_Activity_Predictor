```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("rawDataSerotoninTransporter.csv", sep=";")
df.head()
```

```python
df.info()
```

```python
df = df[df['Standard Relation'] == "'='"].copy()
df["Standard Relation"].unique()
```

Keeping only important features

```python
numeric_columns = df.select_dtypes(include=['float64', 'int64']).columns
df[numeric_columns].corr()
```

```python
important = ["Molecule ChEMBL ID", "Smiles", "Standard Type", "Standard Value", "pChEMBL Value", "Molecular Weight", "AlogP"]
df = df[important]
df.head()
```

Handeling null values

```python
df.info()
```

```python
df_noNull = df.dropna()
df_noNull.info()
```

Creating label using pChEMBL value

```python
df_filtered = df_noNull.copy()
df_filtered["label"] = (df_noNull['pChEMBL Value'] >= 6).astype(int)
df_filtered.head()
```

```python
df_filtered.info()
```

```python
df_filtered["Smiles"].nunique()
```

```python
df_filtered = df_filtered.drop_duplicates(subset="Smiles", keep="first")
df_filtered.info()
```

```python
df_filtered.to_csv("Cleaned_SERT_Data.csv",index=False)
```
